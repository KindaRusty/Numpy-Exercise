#### 20. Consider a (6,7,8) shape array, what is the index (x,y,z) of the 100th element? (★☆☆)

````python
import numpy as np
# Nó thực hiện các phép chia lấy nguyên (floor division) và chia lấy dư (modulo) liên tiếp để "bóc tách" vị trí của phần tử đó trong từng chiều.
#Nhận vào chỉ số phẳng i và cấu trúc mảng (d0,d1,d2,...,dn) và trả về một tuple các chỉ số (i0,i1,i2,...,in) sao cho i = i0*d1*d2*...*dn + i1*d2*...*dn + ... + in-1*dn + in.
#ví dụ: b1 99//8 = 12 dư 3 -> z = 3, tiếp tục 12//7 = 1 dư 5 -> y = 5, tiếp tục 1//6 = 0 dư 1 -> x = 1, nên kết quả là (1,5,3)
a = np.unravel_index(99, (6,7,8)) 
# Sẽ ra kết quả như này
#(np.int64(1), np.int64(5), np.int64(3))
index = tuple(i.item() for i in a) #Chuyển đổi các phần tử trong tuple từ np.int64 sang int
print(a)
print(index)
````

#### 22. Normalize a 5x5 random matrix (★☆☆)

````python
import numpy as np 
#Cách đơn giản nhất là chuẩn hóa về khoảng [0,1] bằng công thức: x_norm = (x - x_min)/(x_max - x_min)
a = np.random.random((5,5))
x_min = a.min()
x_max = a.max()
x_norm = (a - x_min)/(x_max - x_min) #Chuẩn hóa về khoảng [0,1]
print(a)
print(x_norm)
#Cách tối ưu nhất: Linh hoạt và dễ tùy biến, tốc độ nhanh nhất
Z = np.random.random((5, 5))
# Sử dụng keepdims=True để đảm bảo số chiều của mảng không bị giảm
# Điều này giúp phép trừ diễn ra chính xác mà không bị lỗi broadcast
Z = (Z - Z.mean(keepdims=True)) / Z.std(keepdims=True)

print(Z)
````

#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆)

````python
import numpy as np
today = np.datetime64('now', 'D') #np.datetime64() tạo ra một đối tượng datetime64 đại diện cho ngày hiện tại. 'D' là đơn vị thời gian (day) được chỉ định, có nghĩa là chỉ lấy phần ngày mà không quan tâm đến giờ, phút, giây.
yesterday = today - np.timedelta64(1, 'D') #np.timedelta64() tạo ra một khoảng thời gian (timedelta) với số lượng và đơn vị thời gian được chỉ định. Trong trường hợp này, 1 là số lượng và 'D' là đơn vị thời gian (ngày). Vì vậy, np.timedelta64(1, 'D') đại diện cho một khoảng thời gian là 1 ngày.
tomorrow = today + np.timedelta64(1, 'D') #Tương tự như yesterday nhưng cộng thêm 1 ngày để ra ngày mai
print("Today: ", today)
print("Yesterday: ", yesterday)
print("Tomorrow: ", tomorrow)
````

#### 38. Consider a generator function that generates 10 integers and use it to build an array (★☆☆)

````python
import numpy as np
# Tối ưu nhất nếu đã vectorization ngay từ đầu 
gen = np.random.randint(1,10,10) #Lấy 10 số trong khoảng từ 1-10
print(gen)
# Nếu không vectorization dữ liệu ngay từ đâu được (vd: dữ liệu từ api, dòng stream) thì np.fromiter dùng được
def generate():
    for x in range(10):
        yield x #Trong numpy ví dụ có 1 tỷ dữ liệu thì output nó ra từng cái 1 để xử lí thay vì lấy hết ra như khi dùng return (lấy 1 tỷ ouutput ra cùng lúc)
Z = np.fromiter(generate(),dtype=float,count=-1)
print(Z)
````

#### 44. Consider a random 10x2 matrix representing cartesian coordinates, convert them to polar coordinates (★★☆)

````python
import numpy as np
rng = np.random.default_rng()

cartesian = rng.uniform(low=1, high=10, size=(10,2)) #Gen random hệ tọa độ đề-các x,y 10x2
print(cartesian)

#x,y = cartesian[:,0], cartesian[:,1]
#Tính bán kính r (khoảng cách từ gốc tọa độ)
#radius = np.hypot(x,y)
#Tính gốc theta (tính bằng radian)
#theta = np.arctan2(y,x)

#Có thể tính trực tiếp R và Theta nhanh hơn phần trên
radius = np.hypot(cartesian[:, 0], cartesian[:, 1]) #Dùng thay np.sqrt để tối ưu tránh tràn số (overflow) hoặc mất độ chính xác(underflow)
theta = np.arctan2(cartesian[:, 1], cartesian[:, 0]) #Dùng thay arctan vì nó xét đúng dấu của x và y để đặt vào đúng góc phần tư tương ứng và xử lí an trong khi x=0

polar_coords = np.column_stack((radius,theta)) #chạy nhanh hơn một chút so với np.c_[r, theta] khi xử lý các mảng dữ liệu lớn, vì np.c_ sử dụng cơ chế slicing đặc biệt (__getitem__) của Python nên có thêm một chút overhead.
print("Tọa độ cực (r,theta):\n", polar_coords)
````

#### 47. Given two arrays, X and Y, construct the Cauchy matrix C (Cij =1/(xi - yj)) (★★☆)

````python
import numpy as np
rng = np.random.default_rng()
n = 3

# 1. Tạo dữ liệu X và Y là các mảng gồm n giá trị khác nhau
# (Dùng ngẫu nhiên không trùng lặp để tránh x_i == y_j gây chia cho 0)
X_vals = rng.uniform(1, 10, size=n)
Y_vals = rng.uniform(11, 20, size=n)

# 2. Định nghĩa Structured Array
dtype = [('x', float), ('y', float)]

arr_x = np.zeros(n, dtype=dtype)
arr_x['x'] = X_vals
arr_y = np.zeros(n, dtype=dtype)
arr_y['y'] = Y_vals

# 3. Tối ưu bằng Broadcasting: (N, 1) - (N,) -> Tạo Ma trận (N, N)
C = 1 / (arr_x['x'][:, np.newaxis] - arr_y['y']) 
#np.newaxis để thêm cột vào mảng (1d->2d)
#cách đặt np.newaxis ảnh hưởng tới chiều được thêm vào
#A[:, np.newaxis]	(3, 1)	Mảng Cột	Lấy 3 phần tử làm 3 hàng, thêm 1 cột mới.
#A[np.newaxis, :]	(1, 3)	Mảng Hàng	Thêm 1 hàng mới, lấy 3 phần tử làm 3 cột.
print("Ma trận Cauchy C (shape:", C.shape, "):\n", C)
````

````python
import numpy as np
rng = np.random.default_rng()
random = rng.integers(1,100, size=(100,2))
diff = random[:, np.newaxis, :] - random[np.newaxis, :, :]
# random[:, np.newaxis, :] thành shape (100,1,2) có 100 hàng mỗi hàng 1 điểm 1 điểm 2 tọa độ
# random[np.newaxis,:,:] thành shape (1,100,2) có 1 hàng duy nhất chứa 100 điểm, mỗi điểm 2 tọa độ
distance = np.sqrt((diff**2).sum(axis=-1))
print("Size", distance.shape)
# Áp dụng trong K-means Clustering gom nhóm, K-Nearest Neighbors (KNN)
````
